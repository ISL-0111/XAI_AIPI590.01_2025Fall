# Import Dirt_Cleaning.png Image Example

This notebook demonstrates multiple ways to import and display the `Dirt_Cleaning.png` image from the Artifacts folder.

The image shows the **AI Alignment: Cleaning Robot Challenge** interface.

## Method 1: Using PIL (Pillow)

PIL (Python Imaging Library) is a popular library for image processing.

```python
from PIL import Image
import matplotlib.pyplot as plt

# Load the image using PIL
img_pil = Image.open('Artifacts/Dirt_Cleaning.png')

# Display the image
plt.figure(figsize=(12, 8))
plt.imshow(img_pil)
plt.axis('off')
plt.title('AI Alignment: Cleaning Robot Challenge (using PIL)')
plt.tight_layout()
plt.show()

# Print image information
print(f"Image size: {img_pil.size}")
print(f"Image mode: {img_pil.mode}")
print(f"Image format: {img_pil.format}")
```

## Method 2: Using Matplotlib

Matplotlib can directly read and display images.

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# Load the image using matplotlib
img_mpl = mpimg.imread('Artifacts/Dirt_Cleaning.png')

# Display the image
plt.figure(figsize=(12, 8))
plt.imshow(img_mpl)
plt.axis('off')
plt.title('AI Alignment: Cleaning Robot Challenge (using Matplotlib)')
plt.tight_layout()
plt.show()

# Print image information
print(f"Image shape: {img_mpl.shape}")
print(f"Image data type: {img_mpl.dtype}")
```

## Method 3: Using OpenCV (cv2)

OpenCV is a powerful computer vision library. Note that OpenCV reads images in BGR format by default.

```python
import cv2
import matplotlib.pyplot as plt

# Load the image using OpenCV
img_cv = cv2.imread('Artifacts/Dirt_Cleaning.png')

# Convert BGR to RGB for proper display with matplotlib
img_cv_rgb = cv2.cvtColor(img_cv, cv2.COLOR_BGR2RGB)

# Display the image
plt.figure(figsize=(12, 8))
plt.imshow(img_cv_rgb)
plt.axis('off')
plt.title('AI Alignment: Cleaning Robot Challenge (using OpenCV)')
plt.tight_layout()
plt.show()

# Print image information
print(f"Image shape: {img_cv.shape}")
print(f"Image data type: {img_cv.dtype}")
```

## Method 4: Using IPython Display

IPython provides a simple way to display images directly in Jupyter notebooks.

```python
from IPython.display import Image as IPImage, display

# Display the image using IPython
display(IPImage(filename='Artifacts/Dirt_Cleaning.png'))
```

## Working with the Image (Example Analysis)

Once imported, you can perform various operations on the image.

```python
from PIL import Image
import numpy as np
import matplotlib.pyplot as plt

# Load the image
img = Image.open('Artifacts/Dirt_Cleaning.png')

# Convert to numpy array for analysis
img_array = np.array(img)

# Create a figure with subplots
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Original image
axes[0, 0].imshow(img)
axes[0, 0].axis('off')
axes[0, 0].set_title('Original Image')

# Grayscale version
img_gray = img.convert('L')
axes[0, 1].imshow(img_gray, cmap='gray')
axes[0, 1].axis('off')
axes[0, 1].set_title('Grayscale Version')

# Image histogram (if RGB)
if len(img_array.shape) == 3 and img_array.shape[2] >= 3:
    axes[1, 0].hist(img_array[:, :, 0].ravel(), bins=256, color='red', alpha=0.5, label='Red')
    axes[1, 0].hist(img_array[:, :, 1].ravel(), bins=256, color='green', alpha=0.5, label='Green')
    axes[1, 0].hist(img_array[:, :, 2].ravel(), bins=256, color='blue', alpha=0.5, label='Blue')
    axes[1, 0].set_title('Color Histogram')
    axes[1, 0].set_xlabel('Pixel Value')
    axes[1, 0].set_ylabel('Frequency')
    axes[1, 0].legend()

# Resized version
img_resized = img.resize((img.width // 2, img.height // 2))
axes[1, 1].imshow(img_resized)
axes[1, 1].axis('off')
axes[1, 1].set_title('Resized (50%)')

plt.tight_layout()
plt.show()

print(f"\nImage Statistics:")
print(f"Width: {img.width} pixels")
print(f"Height: {img.height} pixels")
print(f"Total pixels: {img.width * img.height:,}")
print(f"Array shape: {img_array.shape}")
```
